```python
import itertools
from itertools import *

list1=[1,2,34,34,56]

sum=itertools.accumulate(list1)
print(list(sum))
```

```python
import itertools
from itertools import *
import operator
list1=[1,2,34,34,56]

mul=itertools.accumulate(list1,operator.mul)
print(list(mul))
```
