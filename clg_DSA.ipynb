```python
# arr = [10, 20, 30, 40, 50]

# Write a function that returns the sum of every pair of adjacent elements.

# Expected output:

# [30, 50, 70, 90]

def sum_of_pair(arr):
    n=len(arr)
    result=[]
    for i in range (n-1):
        sum_pair=arr[i]+arr[i+1]
        result.append(sum_pair)
    return result

arr = [10, 20, 30, 40, 50]
print(sum_of_pair(arr))
```

```python
class Solution(object):
    def maxArea(self, height):
        """
        :type height: List[int]
        :rtype: int
        """
        max=0
        for i in range (0,len(height)):
            for j in range (i+1,len(height)):
                minimum=min(height[i],height[j])
                dist=j-i
                newmax=minimum*dist
                if newmax > max:
                    max=newmax
        return max

arr=[1,8,6,2,5,4,8,3,7]
print(Solution().maxArea(arr))
        
```

```python
def isoef(n,nums):
    n=int(n)
    count=0
    for i in range(len(nums)):
        if nums[i]=='1':
            count+=1
    if count>0:
        print("Hard")
    else:
        print("Easy")
print(isoef(3, ['0', '0', '1']))
```

```python

def isMiddleElementUnique(nums: list[int]) -> bool:
    middle=nums[len(nums)//2]
    found=0
    for num in nums:
        if middle==num:
            found+=1
        if found>1:
            return False
    return True

nums=[1,2]
isMiddleElementUnique(nums)
                
```

```python
def maxValidPairSum(nums: list[int], k: int) -> int:
    n=len(nums)
    #max sum of valid pairs
    max_sum=0
    for i in range(0,len(nums)):
        for j in range(i+1,n):
            if nums[i]+nums[j]<=k:
                max_sum=max(max_sum,nums[i]+nums[j])

nums=[1,3,5,2,8]
k=5
maxValidPairSum(nums, k)
```
