```python
def pattern1(n):
    for i in range(n):
        for j in range(i + 1):
            print("*", end="")
        print()
def pattern2(n):
    for i in range(n):
        for j in range(n - i):
            print("*", end="")
        print()

n = 5
print("Pattern 1:")
pattern1(n)
print("\nPattern 2:")   
pattern2(n)

    
```

```python
def pattern3(n):
    for i in range(n):
        for j in range(i + 1):
            print("*", end="")
        print()
    for i in range(n - 1):
        for j in range(n - i - 1):
            print("*", end="")
        print()
print("\nPattern 3:")
pattern3(n)
```

```python
def pattern4(rows):
    for i in range(1, rows + 1):
        for j in range(1, rows + 1):
            print(j, end=" ")
        print()

print("\nPattern 4:")
pattern4(5)
```

```python
arr = [1, 2, 3, 4, 5]
def alternative(arr):
    new = []
    for i in range(0, len(arr), 2):
        new.append(arr[i])
    return new
print(alternative(arr))
```
