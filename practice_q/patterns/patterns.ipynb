```python
def pattern1(n):
    for i in range(n):
        for j in range(i + 1):
            print("*", end="")
        print()
def pattern2(n):
    for i in range(n):
        for j in range(n - i):
            print("*", end="")
        print()

n = 5
print("Pattern 1:")
pattern1(n)
print("\nPattern 2:")   
pattern2(n)

    
```

```python
def pattern3(n):
    for i in range(n):
        for j in range(i + 1):
            print("*", end="")
        print()
    for i in range(n - 1):
        for j in range(n - i - 1):
            print("*", end="")
        print()
print("\nPattern 3:")
pattern3(n)
```
