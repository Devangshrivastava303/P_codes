```python
def pattern1(n):
    for i in range(n):
        for j in range(i + 1):
            print("*", end="")
        print()
def pattern2(n):
    for i in range(n):
        for j in range(n - i):
            print("*", end="")
        print()

n = 5
print("Pattern 1:")
pattern1(n)
print("\nPattern 2:")   
pattern2(n)

    
```

```python
def pattern3(n):
    for i in range(n):
        for j in range(i + 1):
            print("*", end="")
        print()
    for i in range(n - 1):
        for j in range(n - i - 1):
            print("*", end="")
        print()
print("\nPattern 3:")
pattern3(n)
```

```python
def pattern4(rows):
    for i in range(1, rows + 1):
        for j in range(1, rows + 1):
            print(j, end=" ")
        print()

print("\nPattern 4:")
pattern4(5)
```

```python
def number_of_digits(n):
    count = 0
    while n > 0:
        n //= 10  # Integer division by 10 to remove the last digit
        count += 1 
    return count

print(number_of_digits(12345))
```

```python
str1 = "absd"
str2 = "abcde"
k = 2

# Convert str1 to a list to allow modification
str1_list = list(str1)
length = 0

# Step 1: Fix mismatches using up to k operations
for i in range(len(str1_list)):
    # Ensure we don't go out of bounds if str2 is shorter
    if i >= len(str2):
        break
        
    if str1_list[i] != str2[i] and k > 0:
        str1_list[i] = str2[i]
        k -= 1

# Step 2: Count the total matches after modifications
for i in range(len(str1_list)):
    if i < len(str2) and str1_list[i] == str2[i]:
        length += 1

print(length)   
```

```python
def rotate(nums,k):
    ans=[]
    n=len(nums)
    for i in range(n-k,len(nums)):
        ans.append(nums[i])
    for j in range(0,n-k):
        ans.append(nums[j])
    return ans 
rotate(nums=[1,2,3,4,5,6]
,k=3)
```

```python
def dfs(graph,start,visited):
    visited.add(start)
    for next_node in graph[start]:
        if next_node not in visited:
            dfs(graph,next_node,visited)

graph={
        'A': set(['B','C']),
        'B': set(['A', 'D', 'E']),
        'C': set(['A']),
        'D': set(['B']),
        'E': set(['B']),
    }
visited=set()
dfs(graph,'A',visited)
```
