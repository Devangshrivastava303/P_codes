```python
def sum_Nnum(n):
    if n==0:
        return 0
    
    return n + sum_Nnum(n-1)

n=5
sum_Nnum(n)
```

```python
def fact(n):
    if n==0:
        return 1
    return n*fact(n-1)

n=5
fact(n)
```
