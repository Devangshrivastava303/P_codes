```python
def reverse(n, revers_num=0):
    if n==0:
        return revers_num
    first_digit= n%10   
    revers_num= revers_num*10 + first_digit  # store by 
    num=(n//10)
    return reverse(num, revers_num)

    
    
n=12345
print(reverse(n))
    
```

```python
def power_of_num(n,power):
    if power==0:
        return 1
    return n*power_of_num(n,power-1)
n= 2
power=4
power_of_num(n,power)
```

```python
def fibonacci(n):
    if n==0 :
     return 0
    elif n==1:
       return 1
    else:
       return fibonacci(n-1)+fibonacci(n-2)
    
fibonacci(55)
```
