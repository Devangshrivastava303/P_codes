```python
def binary_search(arr,target):
    if arr==[]:
        return False
    mid=len(arr)//2
    if arr[mid]>target:
        return binary_search(arr[:mid],target)
    if arr[mid]<target:
        return binary_search(arr[mid+1:],target)
    return True
print(binary_search([1,2,4,6,4,2],3))
```

```python
class Solution:
    def sumDigits(self, n):
        return self.helper(n, 0)
    
    def helper(self, n, total):
        if n == 0:
            return total
        return self.helper(n // 10, total + n % 10)
print(Solution().sumDigits(56))
```

```python
class Solution:
    def candy(self, ratings):
        for i in range(len(ratings)):
            if ratings[i] == 0:
                ratings[i] = 1
            if ratings[i] > ratings[i-1]:
                ratings[i] = ratings[i-1] + 1
        return ratings
            
candy = Solution()
print(candy.candy([1,0,2]))
```

```python
def pattern(self, n):
    ans=[]
    if n<0:
        return [n]
    self.recurse(n,n,ans)
    return ans
def recurse(self,n,original,ans):
    ans.append(n)
    if n == 0:
        self.helper(0,original,ans)   
        return                          
    self.recurse(n-5,original,ans)  
def helper(self,n,original,ans):
    if n==original:
        return                          
    n = n+5
    ans.append(n)
    self.helper(n, original, ans)
pattern(10)
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def swapPairs(self, head):
        if not head or not head.next:
            return head
        first=head
        nextt=head.next
        first.next=self.swapPairs(nextt.next) 
        nextt.next=first
        return nextt  

head = ListNode(1, ListNode(2, ListNode(3, ListNode(4))))
result = Solution().swapPairs(head)

while result:
    print(result.val, end=" ")
    result = result.next
```

```python
def reverseExponentiation(self, n):
    if n<10:
        return self.recurse(n,n,0)
    else:
        last_digit=n%10
        first_digit=n//10
        u=last_digit+first_digit
        return self.recurse(n,u,0)
def recurse(self,n,u,i,ans=1):
        if i==u:
            return ans
        return self.recurse(n,u,i+1,ans*n)
reverseExponentiation(23)
```
