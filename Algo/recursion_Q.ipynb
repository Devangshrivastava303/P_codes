```python
def minimum(arr,n):
    if n==1:
        return arr[0]
    
    return min(arr[n-1],minimum(arr,n-1))
print(minimum([5,1,3,2,4],5))   
```

```python
def isvalidpalidrome(s):
    if len(s)==1:
        return True
    if s[0]!=s[len(s)-1]:
        return False
    return isvalidpalidrome(s[1:-1])

print(isvalidpalidrome("racecar"))
```

```python
def sumlist(arr):
    if len(arr) == 0:
        return 0
    return arr[0] + sumlist(arr[1:])

print(sumlist([1, 2, 3, 4]))
```

```python
def occurence(arr,target):
    if len(arr)==0:
        return 0
    if arr[0]==target:
        return 1+occurence(arr[1:],target)
    else:
        return 0+occurence(arr[1:],target)
    
print(occurence([3,2,1,5,6,7,7,8,8,9],5))
```

```python
def reversestr(s):
    if len(s)<=1:
        return s
    return reversestr(s[1:])+s[0]
print(reversestr("how"))
```

```python
def palindrome(s):
    if len(s)<=1:
        return True
    if s[0]!=s[-1]:
        return False
    return palindrome(s[1:-1])
print(palindrome("acecar"))
```

```python
def fib(n):
    if  n==0:
        return 0
    if n==1:
        return 1
    return fib(n-1)+fib(n-2)
print(fib(5))
```

```python
def binary_search(arr,target):
    if arr==[]:
        return False
    mid=len(arr)//2
    if arr[mid]>target:
        return binary_search(arr[:mid],target)
    if arr[mid]<target:
        return binary_search(arr[mid+1:],target)
    return True
print(binary_search([1,2,4,6,4,2],3))
```
