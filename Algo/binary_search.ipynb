```python
def find_first_and_last(arr, target):
    
    def first_element():
        left, right = 0, len(arr) - 1
        while left <= right:
            mid = left + (right - left) // 2
            if arr[mid] >= target:
                right = mid - 1
            else:
                left = mid + 1
        return left

    def last_element():
        left, right = 0, len(arr) - 1
        while left <= right:
            mid = left + (right - left) // 2
            if arr[mid] <= target:
                left = mid + 1
            else:
                right = mid - 1
        return right

    first = first_element()
    last = last_element()

    if first <= last:
        return [first, last]
    return [-1, -1]

arr = [1, 2, 3, 4, 5, 5, 5, 6, 7, 8, 9]
target = 10
print(find_first_and_last(arr, target))
```
