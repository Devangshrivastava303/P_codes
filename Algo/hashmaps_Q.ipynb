```python
my_list=[1,2,3,4,5,234,234,23,4,2344]

element_count={}

for elements in my_list:
    if elements in element_count:
        element_count[elements]+=1
    else:
        element_count[elements]=1
for elements, count in element_count.items():
    print(f"{elements}:{count}")
```

```python
import heapq

arr=[1,4,5,7,3,8,-11]
m=len(arr)

for i in range (m):
    m[i]=-m[i]

heapq.heapify(arr)
def heap_sort(arr):
    heapq.heapify(arr)
    n=len(arr)
    new_list=[0]*n

    for i in range(n):
        minn=heapq.heappop(arr)
        new_list[i]=minn

    return new_list

heap_sort([1,4,5,7,3,8,-11])


```
